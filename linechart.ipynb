```r
months <- c("Jan", "Feb", "Mar", "Apr", "May")
sales <- c(100, 120, 150, 130, 170)
plot(sales,
     type = "l",
     xaxt = "n",
     main = "Line Chart",
     xlab = "Months",
     ylab = "Sales")
axis(1, at = 1:5, labels = months)
```
